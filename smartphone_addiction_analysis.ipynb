#XGBoost

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report
import xgboost as xgb
import matplotlib.pyplot as plt


# Categorize addiction levels
def addiction_category(x):
    if x >= 8:
        return "High"
    elif 5 <= x <= 7:
        return "Moderate"
    elif 3 <= x <= 4:
        return "Low"
    else:
        return "Not Addicted"

# Apply categorization
df["Addiction_Category"] = df["Addiction_Level"].apply(addiction_category)

# Features list (added Gender)
features = [
    "Gender", "Daily_Usage_Hours", "Sleep_Hours", "Academic_Performance",
    "Social_Interactions", "Exercise_Hours", "Anxiety_Level",
    "Depression_Level", "Self_Esteem", "Parental_Control",
    "Screen_Time_Before_Bed", "Phone_Checks_Per_Day", "Apps_Used_Daily",
    "Time_on_Social_Media", "Time_on_Gaming", "Time_on_Education",
    "Phone_Usage_Purpose", "Family_Communication", "Weekend_Usage_Hours"
]

X = df[features].copy()
y = df["Addiction_Category"]

# Encode categorical feature: Phone_Usage_Purpose
le_purpose = LabelEncoder()
X["Phone_Usage_Purpose"] = le_purpose.fit_transform(X["Phone_Usage_Purpose"])

# Encode Gender using the gender_map
gender_map = {"Male": 0, "Female": 1, "Other": 2}
X["Gender"] = X["Gender"].map(gender_map)

# Encode target variable
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

# Train XGBoost model
model = xgb.XGBClassifier(
    use_label_encoder=False, eval_metric="mlogloss", random_state=42
)
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=le_target.classes_))
```

##ROC

```python

from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import accuracy_score, classification_report, roc_curve, auc



# Binarize target for ROC
y_bin = label_binarize(y_encoded, classes=np.unique(y_encoded))
n_classes = y_bin.shape[1]

# Train-test split

X_train, X_test, y_train, y_test = train_test_split(
    X, y_bin, test_size=0.2, random_state=42, stratify=y_bin
)


# Train XGBoost

model = xgb.XGBClassifier(
    use_label_encoder=False, eval_metric="mlogloss", random_state=42
)
model.fit(X_train, y_train)

# Predictions

y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)

print("Accuracy:", accuracy_score(y_test.argmax(axis=1), y_pred.argmax(axis=1)))
print("\nClassification Report:\n", classification_report(y_test.argmax(axis=1), y_pred.argmax(axis=1), target_names=le_target.classes_))

# ROC Curve

fpr, tpr, roc_auc = {}, {}, {}
for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test[:, i], y_proba[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Plot ROC curves
plt.figure(figsize=(8, 6))
for i in range(n_classes):
    plt.plot(fpr[i], tpr[i], lw=2,
             label=f"{le_target.classes_[i]} (AUC = {roc_auc[i]:.2f})")

plt.plot([0, 1], [0, 1], "k--", lw=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - XGBoost (Multi-class)")
plt.legend(loc="lower right")
plt.show()
```

##k cross

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import xgboost as xgb

# Categorize addiction levels
def addiction_category(x):
    if x >= 8:
        return "High"
    elif 5 <= x <= 7:
        return "Moderate"
    elif 3 <= x <= 4:
        return "Low"
    else:
        return "Not Addicted"

df["Addiction_Category"] = df["Addiction_Level"].apply(addiction_category)

# Feature selection
features = [
    "Gender", "Daily_Usage_Hours", "Sleep_Hours", "Academic_Performance",
    "Social_Interactions", "Exercise_Hours", "Anxiety_Level",
    "Depression_Level", "Self_Esteem", "Parental_Control",
    "Screen_Time_Before_Bed", "Phone_Checks_Per_Day", "Apps_Used_Daily",
    "Time_on_Social_Media", "Time_on_Gaming", "Time_on_Education",
    "Phone_Usage_Purpose", "Family_Communication", "Weekend_Usage_Hours"
]
X = df[features].copy()
y = df["Addiction_Category"]

# Encode categorical features
le_purpose = LabelEncoder()
X["Phone_Usage_Purpose"] = le_purpose.fit_transform(X["Phone_Usage_Purpose"])
gender_map = {"Male": 0, "Female": 1, "Other": 2}
X["Gender"] = X["Gender"].map(gender_map)

# Encode target labels
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)

# Standardize features (optional for XGB, but helps sometimes)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Initialize XGBoost
model = xgb.XGBClassifier(use_label_encoder=False, eval_metric="mlogloss", random_state=42)

# Stratified K-Fold
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Store results
all_reports = []
all_accuracies = []

fold = 1
for train_idx, test_idx in kfold.split(X_scaled, y_encoded):
    X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
    y_train, y_test = y_encoded[train_idx], y_encoded[test_idx]

    # Train model
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    # Accuracy
    acc = accuracy_score(y_test, y_pred)
    all_accuracies.append(acc)

    # Classification report
    report = classification_report(y_test, y_pred, target_names=le_target.classes_, output_dict=True)
    all_reports.append(report)

    print(f"\n--- Fold {fold} ---")
    print("Accuracy:", acc)
    print("Classification Report:\n", classification_report(y_test, y_pred, target_names=le_target.classes_))



# Average scores
avg_report = {}
for label in le_target.classes_:
    avg_report[label] = {
        "precision": np.mean([rep[label]["precision"] for rep in all_reports]),
        "recall":    np.mean([rep[label]["recall"] for rep in all_reports]),
        "f1-score":  np.mean([rep[label]["f1-score"] for rep in all_reports])
    }

# Avg accuracy
avg_accuracy = np.mean(all_accuracies)

print("\n Average Classification Report Across Folds ")
for label, scores in avg_report.items():
    print(f"{label:15s} -> Precision: {scores['precision']:.3f}, Recall: {scores['recall']:.3f}, F1: {scores['f1-score']:.3f}")

print(f"\nAverage Accuracy Across Folds: {avg_accuracy:.3f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler, label_binarize
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc
from sklearn.neighbors import KNeighborsClassifier
from collections import defaultdict

# Categorize addiction levels
def addiction_category(x):
    if x >= 8:
        return "High"
    elif 5 <= x <= 7:
        return "Moderate"
    elif 3 <= x <= 4:
        return "Low"
    else:
        return "Not Addicted"

# Apply categorization
df["Addiction_Category"] = df["Addiction_Level"].apply(addiction_category)

# Features & target
features = [
    "Gender", "Daily_Usage_Hours", "Sleep_Hours", "Academic_Performance",
    "Social_Interactions", "Exercise_Hours", "Anxiety_Level",
    "Depression_Level", "Self_Esteem", "Parental_Control",
    "Screen_Time_Before_Bed", "Phone_Checks_Per_Day", "Apps_Used_Daily",
    "Time_on_Social_Media", "Time_on_Gaming", "Time_on_Education",
    "Phone_Usage_Purpose", "Family_Communication", "Weekend_Usage_Hours"
]
X = df[features].copy()
y = df["Addiction_Category"]

# Encode categorical features
le_purpose = LabelEncoder()
X["Phone_Usage_Purpose"] = le_purpose.fit_transform(X["Phone_Usage_Purpose"])
gender_map = {"Male": 0, "Female": 1, "Other": 2}
X["Gender"] = X["Gender"].map(gender_map)

# Encode target labels
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)
class_names = le_target.classes_

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# K-Fold setup
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
knn = KNeighborsClassifier(n_neighbors=5)

all_reports = []
all_conf_matrices = []
all_probs = []
all_true = []

for train_idx, test_idx in kf.split(X_scaled, y_encoded):
    X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
    y_train, y_test = y_encoded[train_idx], y_encoded[test_idx]

    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    y_prob = knn.predict_proba(X_test)

    # Collect results
    all_reports.append(classification_report(y_test, y_pred, target_names=class_names, output_dict=True))
    all_conf_matrices.append(confusion_matrix(y_test, y_pred))
    all_probs.append(y_prob)
    all_true.append(y_test)

# ==== Average Classification Report ====
# Average scores
avg_report = {}
for label in le_target.classes_:
    avg_report[label] = {
        "precision": np.mean([rep[label]["precision"] for rep in all_reports]),
        "recall":    np.mean([rep[label]["recall"] for rep in all_reports]),
        "f1-score":  np.mean([rep[label]["f1-score"] for rep in all_reports])
    }

# Avg accuracy
avg_accuracy = np.mean(all_accuracies)

print("\n Average Classification Report Across Folds ")
for label, scores in avg_report.items():
    print(f"{label:15s} -> Precision: {scores['precision']:.3f}, Recall: {scores['recall']:.3f}, F1: {scores['f1-score']:.3f}")

print(f"\nAverage Accuracy Across Folds: {avg_accuracy:.3f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from catboost import CatBoostClassifier

# Categorize addiction levels
def addiction_category(x):
    if x >= 8:
        return "High"
    elif 5 <= x <= 7:
        return "Moderate"
    elif 3 <= x <= 4:
        return "Low"
    else:
        return "Not Addicted"

# Apply categorization
df["Addiction_Category"] = df["Addiction_Level"].apply(addiction_category)

# Features
features = [
    "Gender", "Daily_Usage_Hours", "Sleep_Hours", "Academic_Performance",
    "Social_Interactions", "Exercise_Hours", "Anxiety_Level",
    "Depression_Level", "Self_Esteem", "Parental_Control",
    "Screen_Time_Before_Bed", "Phone_Checks_Per_Day", "Apps_Used_Daily",
    "Time_on_Social_Media", "Time_on_Gaming", "Time_on_Education",
    "Phone_Usage_Purpose", "Family_Communication", "Weekend_Usage_Hours"
]
X = df[features].copy()
y = df["Addiction_Category"]

# Encode categorical features
le_purpose = LabelEncoder()
X["Phone_Usage_Purpose"] = le_purpose.fit_transform(X["Phone_Usage_Purpose"])
gender_map = {"Male": 0, "Female": 1, "Other": 2}
X["Gender"] = X["Gender"].map(gender_map)

# Encode target labels
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)

# Scale numerical features (CatBoost doesn’t need scaling, but it’s fine to keep consistent)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

# Initialize CatBoost
model = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=6,
    loss_function="MultiClass",
    random_state=42,
    verbose=0  # suppress training logs
)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)
y_pred = y_pred.flatten()  # CatBoost returns 2D array

# Accuracy & Report
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=le_target.classes_))

```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report
from catboost import CatBoostClassifier
from collections import defaultdict

# Categorize addiction levels
def addiction_category(x):
    if x >= 8:
        return "High"
    elif 5 <= x <= 7:
        return "Moderate"
    elif 3 <= x <= 4:
        return "Low"
    else:
        return "Not Addicted"

# Apply categorization
df["Addiction_Category"] = df["Addiction_Level"].apply(addiction_category)

# Features
features = [
    "Gender", "Daily_Usage_Hours", "Sleep_Hours", "Academic_Performance",
    "Social_Interactions", "Exercise_Hours", "Anxiety_Level",
    "Depression_Level", "Self_Esteem", "Parental_Control",
    "Screen_Time_Before_Bed", "Phone_Checks_Per_Day", "Apps_Used_Daily",
    "Time_on_Social_Media", "Time_on_Gaming", "Time_on_Education",
    "Phone_Usage_Purpose", "Family_Communication", "Weekend_Usage_Hours"
]
X = df[features].copy()
y = df["Addiction_Category"]

# Encode categorical features
le_purpose = LabelEncoder()
X["Phone_Usage_Purpose"] = le_purpose.fit_transform(X["Phone_Usage_Purpose"])
gender_map = {"Male": 0, "Female": 1, "Other": 2}
X["Gender"] = X["Gender"].map(gender_map)

# Encode target labels
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)
class_names = le_target.classes_

# Scale numerical features (optional for CatBoost)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5-Fold Cross Validation
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

fold = 1
accuracies = []

# For averaging classification report
report_sum = defaultdict(lambda: {"precision":0, "recall":0, "f1-score":0, "support":0})

for train_index, test_index in kf.split(X_scaled, y_encoded):
    X_train, X_test = X_scaled[train_index], X_scaled[test_index]
    y_train, y_test = y_encoded[train_index], y_encoded[test_index]

    # Initialize CatBoost
    model = CatBoostClassifier(
        iterations=500,
        learning_rate=0.05,
        depth=6,
        loss_function="MultiClass",
        random_state=42,
        verbose=0
    )

    # Train
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test).flatten()

    # Accuracy
    acc = accuracy_score(y_test, y_pred)
    accuracies.append(acc)

    # Classification report for this fold
    report = classification_report(y_test, y_pred, target_names=class_names)
    print(f"\nFold {fold} Accuracy: {acc:.4f}")
    print(f"Fold {fold} Classification Report:\n{report}")

    # Accumulate metrics for average report
    report_dict = classification_report(y_test, y_pred, target_names=class_names, output_dict=True)
    for cls in class_names:
        report_sum[cls]["precision"] += report_dict[cls]["precision"]
        report_sum[cls]["recall"] += report_dict[cls]["recall"]
        report_sum[cls]["f1-score"] += report_dict[cls]["f1-score"]
        report_sum[cls]["support"] += report_dict[cls]["support"]

    fold += 1

# Average Accuracy
print("\nAverage Accuracy over 5 folds:", np.mean(accuracies))

# Average Classification Report
avg_report = {}
print("\nAverage Classification Report over 5 folds:")
for cls in class_names:
    avg_report[cls] = {
        "precision": report_sum[cls]["precision"] / 5,
        "recall": report_sum[cls]["recall"] / 5,
        "f1-score": report_sum[cls]["f1-score"] / 5,
        "support": report_sum[cls]["support"] / 5
    }
    print(f"{cls}: precision={avg_report[cls]['precision']:.4f}, recall={avg_report[cls]['recall']:.4f}, f1-score={avg_report[cls]['f1-score']:.4f}, support={avg_report[cls]['support']:.1f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Manually Parse and Structure the Data from User Input ---
# This data is taken directly from your text descriptions of the model reports.
data = {
    'XGBoost': {
        'High': {'Precision': 0.932, 'Recall': 0.987, 'F1-Score': 0.959},
        'Low': {'Precision': 0.333, 'Recall': 0.133, 'F1-Score': 0.189},
        'Moderate': {'Precision': 0.621, 'Recall': 0.591, 'F1-Score': 0.603},
        'Not Addicted': {'Precision': 0.344, 'Recall': 0.235, 'F1-Score': 0.278}
    },
    'KNN': {
        'High': {'Precision': 0.806, 'Recall': 0.986, 'F1-Score': 0.887},
        'Low': {'Precision': 0.200, 'Recall': 0.033, 'F1-Score': 0.057},
        'Moderate': {'Precision': 0.414, 'Recall': 0.146, 'F1-Score': 0.215},
        'Not Addicted': {'Precision': 0.288, 'Recall': 0.058, 'F1-Score': 0.096}
    },
    'CatBoost': {
        'High': {'Precision': 0.9265, 'Recall': 0.9948, 'F1-Score': 0.9594},
        'Low': {'Precision': 0.1000, 'Recall': 0.0333, 'F1-Score': 0.0500},
        'Moderate': {'Precision': 0.6711, 'Recall': 0.6771, 'F1-Score': 0.6728},
        'Not Addicted': {'Precision': 0.3552, 'Recall': 0.1925, 'F1-Score': 0.2488}
    },
    'LightGBM': {
        'High': {'Precision': 0.9358, 'Recall': 0.9847, 'F1-Score': 0.9596},
        'Low': {'Precision': 0.3167, 'Recall': 0.1333, 'F1-Score': 0.1844},
        'Moderate': {'Precision': 0.6358, 'Recall': 0.6229, 'F1-Score': 0.6277},
        'Not Addicted': {'Precision': 0.3571, 'Recall': 0.2537, 'F1-Score': 0.2955}
    },
    'SVM': {
        'High': {'Precision': 0.806, 'Recall': 0.986, 'F1-Score': 0.887},
        'Low': {'Precision': 0.200, 'Recall': 0.033, 'F1-Score': 0.057},
        'Moderate': {'Precision': 0.414, 'Recall': 0.146, 'F1-Score': 0.215},
        'Not Addicted': {'Precision': 0.288, 'Recall': 0.058, 'F1-Score': 0.096}
    }
}

# --- 2. Create the Radar Charts ---
labels = ['Precision', 'Recall', 'F1-Score']
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1] # This is to close the plot loop

# Create a 2x2 subplot grid for the four classes
fig, axs = plt.subplots(figsize=(14, 14), nrows=2, ncols=2, subplot_kw=dict(polar=True))
axs = axs.flatten()

# Define the classes to plot and the new color palette
classes = ['High', 'Low', 'Moderate', 'Not Addicted']
models = list(data.keys())
colors = ['#3498db', '#9b59b6', '#1abc9c', '#f39c12', '#e74c3c'] # Eyecatching colors

# Loop through each class and create a chart on its own subplot
for ax, class_name in zip(axs, classes):
    # Set the labels for the axes
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels, size=12)

    # Set the title for each subplot
    ax.set_title(f"Performance for '{class_name}' Class", size=16, y=1.12, weight='bold')

    # Set the radial grid lines and labels
    ax.set_rlabel_position(22.5)
    ax.set_yticks([0.2, 0.4, 0.6, 0.8])
    ax.set_yticklabels(["0.2", "0.4", "0.6", "0.8"], color="grey", size=9)
    ax.set_ylim(0, 1)

    # Loop through each model to plot its data on the current subplot
    for i, model_name in enumerate(models):
        values = list(data[model_name][class_name].values())
        values += values[:1] # Close the plot loop

        # Plot the line
        ax.plot(angles, values, color=colors[i], linewidth=2.5, linestyle='solid', label=model_name)
        # Fill the area under the line
        ax.fill(angles, values, color=colors[i], alpha=0.15)

# Add a single, shared legend to the entire figure
handles, labels = ax.get_legend_handles_labels()
fig.legend(handles, labels, loc='upper right', bbox_to_anchor=(1.15, 0.95), fontsize=14)

plt.tight_layout(pad=3.0)
# Use plt.show() to display the plot in a notebook or script
# Use plt.savefig() to save it to a file
plt.savefig('class_performance_radar_charts_cool_colors.png')
```
